### 1. TypedDict

- `Dict` / `TypedDict` 모두 기본값은 타입 검사 안 함
- (VS Code) Pylance 설치 후 `Ctrl + ,` → `type checking mode`를 `off → basic`으로 바꾸면 `TypedDict`만 IDE가 타입 에러를 표시
- 정적 검사라 타입이 틀려도 **실행은 정상** (에러 표시 ≠ 실행 중단)
- `TypedDict`는 키 자동완성 · 값 타입 추론 등 **타입 힌트도 제공** (일반 `Dict`는 키를 몰라 자동완성 불가)

```python
from typing import Dict, TypedDict

# Dict: 값 타입을 하나로만 지정 → 키별 구분 불가
person_dict: Dict[str, str | bool] = {
    "name": "상우",
    "age": "34",   # 값 타입에 str 포함되므로 통과
    "job": "개발자",
    "alive": True,
}

# TypedDict: 키마다 타입 지정 → 구조를 명확히 표현
class Person(TypedDict):
    name: str
    age: int
    job: str
    alive: bool
```

### 2. 생성 & 출력

- 런타임에는 둘 다 그냥 `dict` → 출력 결과 형태는 동일

```python
# Dict
print(person_dict)

# TypedDict 생성
person = Person({"name": "상우", "age": 34, "job": "개발자", "alive": True})
print(person)
```

### 3. 잘못된 타입 대입

- `Dict`: 어떤 값을 넣어도 IDE가 잡지 않음
- `TypedDict`: 타입이 틀리면 IDE가 에러 표시 (단, 실행은 그대로 됨)

```python
# Dict: 타입 틀려도 IDE 경고 없음
person_dict["alive"] = "살아있음"
print(person_dict)

# TypedDict: alive는 bool인데 str 대입 → IDE가 에러 표시 (실행은 됨)
person["alive"] = "살아있음"
print(person)
```
